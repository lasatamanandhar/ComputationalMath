 1. check how the numerical differentiation error changes when different step sizes h and different values of a are used

```python
import numpy as np
import matplotlib.pyplot as plt

def derivative(func, a, method='forward', h=0.001):

    if method == 'forward':
        return (func(a + h) - func(a)) / h

    elif method == 'backward':
        return (func(a) - func(a - h)) / h

    else:
        raise ValueError("Method must be 'forward' or 'backward'.")

# True derivative of cos(x) = -sin(x)

# Different values of h
h_values = [1, 0.1, 0.01, 0.001, 0.0001]

print("Checking Error for Different h values:\n")

for h in h_values:
    actual_value = -np.sin(0.0)

    result = derivative(np.cos, 0, method='forward', h=h)

    error = np.abs(result - actual_value)

    print("h =", h)
    print("Result = {:0.6f}".format(result))
    print("Error  = {:0.6f}".format(error))
    print()

# Different values of a
a_values = [0, np.pi/4, np.pi/2, np.pi]

print("Checking Error for Different a values:\n")

for a in a_values:
    actual_value = -np.sin(a)

    result = derivative(np.cos, a, method='forward', h=0.001)

    error = np.abs(result - actual_value)

    print("a =", a)
    print("Result = {:0.6f}".format(result))
    print("Actual = {:0.6f}".format(actual_value))
    print("Error  = {:0.6f}".format(error))
    print()
```

2. add the Central Difference formula to the derivative function and compares it with Forward and Backward Difference methods

```python
import numpy as np

def derivative(func, a, method='forward', h=0.001):

    if method == 'forward':
        return (func(a + h) - func(a)) / h

    elif method == 'backward':
        return (func(a) - func(a - h)) / h

    elif method == 'central':
        return (func(a + h) - func(a - h)) / (2 * h)

    else:
        raise ValueError("Method must be 'central', 'forward' or 'backward'.")

# Testing at x = 0
a = 0

actual_value = -np.sin(a)

forward_result = derivative(np.cos, a, method='forward')
backward_result = derivative(np.cos, a, method='backward')
central_result = derivative(np.cos, a, method='central')

print("Actual Value =", actual_value)
print()

print("Forward Difference =", forward_result)
print("Forward Error =", abs(forward_result - actual_value))
print()

print("Backward Difference =", backward_result)
print("Backward Error =", abs(backward_result - actual_value))
print()

print("Central Difference =", central_result)
print("Central Error =", abs(central_result - actual_value))
```

3. calculate derivatives for multiple input values using arrays and plots the numerical derivative against the true derivative

```python
import numpy as np
import matplotlib.pyplot as plt

def derivative(func, a, method='central', h=0.001):

    if method == 'forward':
        return (func(a + h) - func(a)) / h

    elif method == 'backward':
        return (func(a) - func(a - h)) / h

    elif method == 'central':
        return (func(a + h) - func(a - h)) / (2 * h)

    else:
        raise ValueError("Method must be 'central', 'forward' or 'backward'.")

# Generate x values
x = np.linspace(0, 5*np.pi, 100)

# Function y = sin(x)
y = np.sin(x)

# Numerical derivative
dy = derivative(np.sin, x, method='central')

# True derivative of sin(x) is cos(x)
true_dy = np.cos(x)

# Plotting
plt.plot(x, dy, label='Numerical Derivative')
plt.plot(x, true_dy, label='True Derivative')

plt.xlabel("x")
plt.ylabel("Derivative")
plt.title("Derivative of sin(x)")
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def derivative(func, a, method='central', h=0.001):

    if method == 'central':
        return (func(a + h) - func(a - h)) / (2 * h)

x = np.linspace(-10, 10, 100)

# Function
y = x**2

# Numerical derivative
dy = derivative(lambda x: x**2, x)

# True derivative
true_dy = 2*x

# Plot
plt.plot(x, dy, label='Numerical Derivative')
plt.plot(x, true_dy, label='True Derivative')

plt.xlabel("x")
plt.ylabel("Derivative")
plt.title("Derivative of x^2")

plt.legend()
plt.show()
```
